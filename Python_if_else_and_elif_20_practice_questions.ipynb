**16. Write a Python program to check whether a triangle is equilateral, isosceles, or scalene.**


```python
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a == b and b == c:                 # all 3 sides are equal.
    print("Equilateral triangle")
elif a == b or b == c or a == c:      # any 2 sides are equal.
    print("Isosceles triangle")
else:
    print("Scalene triangle")         # all 3 sides are different.
```

**18. Write a Python program to create a simple calculator using if, elif, and else for +, -, x, and /
operations.**

```python
a = 10
b = 20
operator = input("Enter operator: ")

if operator == "+":
    print(a + b)
elif operator == "-":
    print(a - b)
elif operator == "*":
    print(a * b)
elif operator == "/":
    print(a / b)
else:
    print("Invalid operator")
```
