### Assignment Operators - 10 Questions 


**1. Create a variable `x = 10.` Add `5` to `x` using an assignment operator and print the final value.** 


```python
x=10
x+=5
print('Final value',x)
```

**2. Create a variable `marks = 80.` Subtract `10` using an assignment operator and print marks.** 


```python
marks=80
marks-=10
print('Final marks',marks)
```

**3. Create a variable `price = 50`. Multiply price by `3` using an assignment operator.** 


```python
price = 50
price *= 3
print('Final Price',price)
```

**4. Create a variable `total = 100`. Divide total by `4` using an assignment operator.** 


```python
total = 100
total /= 4
print('Total',total)
```

**5. Create a variable `num = 17`. Use an assignment operator to store the remainder when divided by `5`.** 


```python
num = 17
num %= 5
print('Remainder',num)
```

**6. Create a variable `a = 2`. Use an assignment operator to calculate a raised to power `3`.** 


```python
a = 2
a **= 3
print('Final Value',a)
```

**7. Create a variable `b = 25`. Use floor division assignment with `4` and print the result.**


```python
b = 25
b //= 4
print('Final Value',b)
```

**8. Take a number from the user and increase it by `10` using `+=`.**


```python
num = int(input("Enter the number"));
num += 10
print('Final num',num)
```

**9. Take a salary from the user and decrease it by `500` using `-=`.** 

```python
salary = int(input("Enter your salary"));
salary -= 500
print('Final salary', salary)
```

**10. Write a program using at least three assignment operators: `+=`, `*=`, and `-=` on one variable.**

```python
num = 10

num += 5
num *= 2
num -= 4

print(num)
```

**9. Write a program to calculate simple interest using (P * R * T) / 100** 

```python
P = float(input("Enter Principal: "))
R = float(input("Enter Rate: "))
T = float(input("Enter Time: "))

SI = (P * R * T) / 100

print("Simple Interest =", SI)
```
