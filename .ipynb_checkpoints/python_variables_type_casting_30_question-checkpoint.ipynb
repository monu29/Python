**5. Swap two variables without third variable.**


```python
a = 14
b = 10

# Add the values of a and b
a += b
print("sum :", a)

#swaping values
b = a-b
a = a-b

print("a :", a)
print("b :", b)
```

**6. Swap two variables using third variable.**


```python
a = 14
b = 12

#assigning value to third variable to third variable
c = b

#swaping values
a = b
b = c

print("a :", a)
print("b :", b)
```
