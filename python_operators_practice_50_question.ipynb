### Assignment 


**2. Use assignment operators `(=, +=, -=, *=, /=, %=, //=, **=)` in a short code snippet.** 


```python
x = 10

x += 5
x -= 2
x *= 3
x /= 2
x %= 4
x //= 2
x **= 3

print(x)
```

### Comparison 

**3. Compare two values using comparison operators and state whether the result is True or False.** 


```python
a = 10
b = 5

print(a == b)   # False
print(a != b)   # True
print(a > b)    # True
print(a < b)    # False
print(a >= b)   # True
print(a <= b)   # False
```

### Logical & Mixed 


**4. Create a small program using logical and comparison operators together.** 


```python
a = 10
b = 5

result = (a > b) and (a != b)

print(result)
```
