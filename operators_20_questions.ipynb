<h1 style="text-align: center;">Python Operators</h1>

### 1. What are operators in Python?

Operators in Python are symbols used to perform operations on values and variables.

**Examples:**  +, -, *, /

### 2. What is the difference between arithmetic and assignment operators?


| Arithmetic Operators | Assignment Operators |
|---|---|
| Used to perform mathematical operations. | Used to assign or update values in variables. |
| `+`, `-`, `*`, `/`, `%`, `**`, `//` | `=`, `+=`, `-=`, `*=`, `/=` |
| Example: `a + b` | Example: `a += 5` |
| Used to calculate values. | Used to store or update values. |

### 3. Write a program to add two numbers using the `+` operator.

```python
a = 10
b = 50
print(a + b)
```

### 4. Write a program to find the remainder using the `%` operator.

```python
a = 17
b = 5
print(a % b)
```

### 5. Explain the difference between / and // operators.

| `/` Operator | `//` Operator |
|---|---|
| Used for normal division. | Used for floor division. |
| Returns a decimal/float value. | Returns the floor value. |
| Example: `10 / 3` = `3.333...` | Example: `10 // 3` = `3` |

### 6. Write a program to calculate the power of a number using `**`.

```python
a = 2
b = 3
print(a ** b)
```

### 7. What is the difference between `==` and `=`?

| `=` | `==` |
|---|---|
| Assignment operator | Comparison operator |
| Used to assign a value to a variable | Used to compare two values |
| Example: `x = 10` | Example: `x == 10` |
| Does not return `True` or `False` | Returns `True` or `False` |

### 8. Write a program to check whether two numbers are equal.

```python
a = 10
b = 10

print(a == b)
```

### 9. Explain the != operator with an example.

The `!=` operator is used to check whether two values are **not equal**.

```python
a = 10
b = 20

print(a != b)
```

### 10. Write a program using > and < operators.

```python
a = 10
b = 20

print(a > b)
print(a < b)
```

### 11. What are logical operators in Python?

**Logical operators** are used to combine two or more conditions.
| Operator | Meaning |
|---|---|
| `and` | True when both conditions are True |
| `or` | True when at least one condition is True |
| `not` | Reverses the result |

### 12. Write a program using and operator.

```python
a = 10

print(a > 5 and a < 20)
```

### 13. Write a program using or operator.

```python
a = 10

print(a > 20 or a < 15)
```

### 14. Write a program using not operator.

```python
a = 10

print(not(a > 20))
```

### 15. What are comparison operators? List all of them.

Comparison operators are used to compare two values. They return either `True` or `False`.


| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

### 16. Explain += and -= with examples.

`+=` adds a value to a variable and updates it. 

```python
x = 10
x += 5
print(x)
```

### 17. What is the use of *= operator?

The `*=` operator is used to multiply a variable by a value and assign the result back to the same variable.

```python
x = 5
x *= 3
print(x)
```

### 18. What are membership operators? Explain in and not in.

**Membership operators** are used to check whether a value is present in a sequence.

| Operator | Meaning | Example |
|---|---|---|
| `in` | Checks if a value is present. | `"a" in "apple"` → `True` |
| `not in` | Checks if a value is not present. | `"b" not in "apple"` → `True` |

```python
text = "Python"
print("P" in text)
print("x" not in text)
```

### 19. What are identity operators? Explain is and is not.

**Identity operators** are used to check whether two variables refer to the same object.

| Operator | Meaning |
|---|---|
| `is` | Checks whether two objects are the same object. |
| `is not` | Checks whether two objects are not the same object. |

```python
a = [1, 2, 3]
b = a
print(a is b)
print(a is not b)
```
